```python
import pandas as pd
from pathlib import Path
import datetime as dt
```

```python
from datetime import date


def fetch_teams_date_sample_submissions(main_folder_path = Path("model-output")):
    """
    Fetches the name and date of reference of each team's sample submissions 
"""
    found_samples = []
    for subfolder_path in main_folder_path.iterdir():
        if subfolder_path.is_dir():
            for file_path in subfolder_path.iterdir():
                if file_path.is_file() and file_path.suffix == ".csv":
                    # first, for each file, filter the column "output_type" for sample, and set the reference_date column to datetime. extract the team's name (which is the name of the subfolder) and the date range from unique values of the reference_date column. append the team's name and the date range to a list of tuples.
                    df = pd.read_csv(file_path)
                    
                    df = df[df["output_type"] == "sample"]
                    df = df.dropna(subset=["reference_date"])
                    df = df.drop_duplicates(subset=["reference_date"])
                    df["reference_date"] = df["reference_date"].astype(str).str.strip()
                    df["reference_date"] = pd.to_datetime(df["reference_date"], errors="coerce", format="%Y-%m-%d")
                    team_name = subfolder_path.name
                    if not df.empty:
                        # the unique value of the date
                        date_ = df["reference_date"].unique()
                        found_samples.append((team_name, date_))
                elif file_path.is_file() and file_path.suffix == ".parquet":
                                    # first, for each file, filter the column "output_type" for sample, and set the reference_date column to datetime. extract the team's name (which is the name of the subfolder) and the date range from unique values of the reference_date column. append the team's name and the date range to a list of tuples.
                                    df = pd.read_parquet(file_path)
                                    
                                    df = df[df["output_type"] == "sample"]
                                    df = df.dropna(subset=["reference_date"])
                                    df = df.drop_duplicates(subset=["reference_date"])
                                    df["reference_date"] = df["reference_date"].astype(str).str.strip()
                                    df["reference_date"] = pd.to_datetime(df["reference_date"], errors="coerce", format="%Y-%m-%d")
                                    team_name = subfolder_path.name
                                    if not df.empty:
                                        # the unique value of the date
                                        date_ = df["reference_date"].unique()
                                        found_samples.append((team_name, date_))

    found_samples_df = pd.DataFrame(found_samples, columns=["team_name", "date_"])
    # how to take hout hours min sec from the date_ column but please each date is not a list!
    found_samples_df["date_"] = found_samples_df["date_"].apply(lambda x: x[0] if len(x) > 0 else pd.NaT)
    # now order by date
    found_samples_df = found_samples_df.sort_values(by="date_").reset_index(drop=True)


    return found_samples_df
```

```python
found_samples_df = fetch_teams_date_sample_submissions(main_folder_path = Path("model-output"))
```
